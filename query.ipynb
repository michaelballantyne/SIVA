```python
# ===== Declarative shallow DSL flow =====
# Pipeline: inspect -> (subset) -> load -> render / compress.
# subset() narrows (metadata only, pure). load() materializes what the info
# describes. render()/compress() act on already-loaded data (raise otherwise).
from my_subset import subset
from my_estimate import estimate_render_cost, format_estimate

dataset = inspect_file("/projects/autonomousvis/ashrestha/code/VisLang/csafe_heptane_302x302x302_uint8.raw")

# 1) Overview — load the whole thing, then render it.
render(load(dataset))

# 2) Big file? Estimate the render cost first, then stride per its recommendation.
print(format_estimate(estimate_render_cost(dataset.filepath)))      # -> recommends {'grid': 255}

# 3) Refine — narrow with subset(), load, then render just that.
view = subset(dataset, dimensions={'grid': 150})                    # 302^3 -> ~151^3
render(load(view))

# 4) Inspect the materialized arrays in hand.
loaded = load(subset(dataset, variables=['csafe_heptane']))
print(loaded)

# 5) compress() is HDF5-only — load first, e.g. on the nyx .hdf5 dataset:
#    h5 = inspect_file(".../NVB_C009_l10n512_S12345T692_z42.hdf5")
#    compressed = compress(load(subset(h5, variables=['density'])), ['density'], error_bound=0.01)
```
